```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
content = df['type'].value_counts()
print(content)
```

```python
conten_percentage = df['type'].value_counts(normalize=True)*100
print(conten_percentage)
```

```python
movies = df[df['type'] == 'Movie']
tv_shows = df[df['type'] == 'TV Show']
```

```python
movies.head()
```

```python
tv_shows.head()
```

```python
total_movies = len(movies)
total_tv_shows = len(tv_shows)
```

```python
total_content = total_movies + total_tv_shows
```

```python
movie_percentage = (total_movies / total_content) * 100
tv_show_percentage = (total_tv_shows / total_content) * 100
```

```python
summary = pd.DataFrame({
    'Content Type': ['movies', 'tv_shows'],
    'Total Count': [total_movies, total_tv_shows],
    'Percentage': [movie_percentage, tv_show_percentage]
})

summary
```

```python
content.plot(kind='bar')

plt.title("Netflix Content Distribution")
plt.xlabel("Content Type")
plt.ylabel("Number of Titles")
plt.show()
```

```python
My_Labels = ['Movie', 'TV Show']
plt.pie(content, labels = My_Labels, shadow = True, autopct = '%1.1f%%')
plt.title('Netflix Content Percentage Distribution')
plt.show()
```

```python
#Key Findings:-

highest_category = summary.loc[
    summary['Total Count'].idxmax()
]

lowest_category = summary.loc[
    summary['Total Count'].idxmin()
]

print("KEY OBSERVATIONS")

print("Total Content:", total_content)

print(
    "Highest Content Category:",
    highest_category['Content Type']
)

print(
    "Highest Category Percentage:",
    round(highest_category['Percentage'], 2), "%"
)

print(
    "Lowest Content Category:",
    lowest_category['Content Type']
)

print(
    "Lowest Category Percentage:",
    round(lowest_category['Percentage'], 2), "%"
)
```
